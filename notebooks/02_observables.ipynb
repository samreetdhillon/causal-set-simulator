# Observables: Dimension, Chains, and Antichains

This notebook computes the main observables used to analyze a causal set.
If `scipy` is unavailable, the dimension estimator is skipped.

```python
from pathlib import Path
import sys
import importlib.util

cwd = Path.cwd()
if (cwd / "src").exists():
    PROJECT_ROOT = cwd
elif (cwd.parent / "src").exists():
    PROJECT_ROOT = cwd.parent
else:
    raise RuntimeError("Could not locate project root (expected a src/ folder).")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

HAS_SCIPY = importlib.util.find_spec("scipy") is not None
print("scipy available:", HAS_SCIPY)
```

```python
import numpy as np
from src.core_operations import sprinkle, causal_matrix

try:
    from src import observables as obs
    HAS_OBS = True
except Exception as exc:
    HAS_OBS = False
    print("Observables import failed (likely missing scipy):", exc)

N = 70
dim = 2
points = sprinkle(N, dim=dim)
R = causal_matrix(points, dim=dim)

if HAS_OBS:
    r = obs.ordering_fraction(R)
    d_est = obs.estimate_dimension(r)
    L = obs.longest_chain_length(R)
    AC = obs.largest_antichain(R)
    print(f"Ordering fraction r: {r:.3f}")
    if d_est is not None:
        print(f"Estimated dimension: {d_est:.2f}")
    else:
        print("Estimated dimension: not available for this r")
    print(f"Longest chain length: {L}")
    print(f"Largest antichain size: {AC}")
else:
    print("Skipping observables. Install scipy to enable them.")
```

## Interpreting the Numbers
- $r$ (ordering fraction) indicates how many pairs are causally related.
- The Myrheim-Meyer dimension inverts a known theoretical curve; in 2D, it should be near 2.
- The longest chain approximates the longest timelike path through the set.
- The largest antichain approximates a spacelike slice with maximal width.
